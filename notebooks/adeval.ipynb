```python
generator_prompt = f"""
Analyze this product page: {http_link}

Generate 3 distinct advertisement concepts for the shoes.

Output only a single JSON array. Do not include any introductory, concluding, or explanatory text.

Each object in the array must have two keys:

"target": A string describing the target market and ad style.

"html": A string containing the complete, self-contained HTML file for a one-page pamphlet.  Use links to display images.
"""
```

```python
print(result.text)
```

```python
import json
result_json = json.loads(result.text)
```

```python
from pathlib import Path
from IPython.display import HTML

brand = "nike"
data_dir = Path(f"../data/ads/{brand}")

for idx in range(len(result_json)):
    file = data_dir / f"{brand}{idx}.html"
    file.write_text(result_json[idx]["html"])
```
